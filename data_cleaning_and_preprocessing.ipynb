```python
import pandas as pd
df = pd.read_csv("data3.csv")
df
```

```python
# to know where the null value is present : true comes there 
df.isnull()
```

```python
# to know how many null values are present in a particular column :
df.isnull().sum() 
```

```python
# only those rows where no null values present : 
df.dropna()
```

```python
# removing those columns which contains missing values : 
df.dropna(axis = 1) 
```

```python
# filling missing values with 0 : 
df.fillna(0) 
```

```python
# for filling the null values with their nearest forward values : 
df.ffill()
```

```python
# filling null values with their nearest backward values : 
df.bfill()
```

```python
# filling null values with the mean values : 
df["Age"].fillna(df["Age"].mean())
```

```python
# loading duplicate data : 
df.duplicated()
```

```python
# dropping dupliacted : 
df.drop_duplicates() 
```
